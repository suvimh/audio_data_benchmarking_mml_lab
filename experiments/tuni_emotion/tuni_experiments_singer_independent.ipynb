## 3. Feature selection with the best model-embedding pairs

Choose the pairs that performed best in the baseline run (section 2). Current
top trad-ML pairs by **Val Balanced Accuracy** from the 500ms run,
`experiments/tuni_emotion/metrics/tuni_emotion_full_metrics_outline.csv`:

| embedding | model | Val Balanced Accuracy |
|-----------|-------|-----------------------|
| opensmile | svm   | 0.362                 |
| clap      | svm   | 0.350                 |


Edit the `pairs` list below to keep the sweep cheap. Each pair runs the full
RFE + SelectKBest sweep over `n_features_list` (counts >= the embedding size
are skipped automatically, e.g. whisper/clap stop at 320).

for the 3 second run `experiments/tuni_emotion/metrics/3sec_data/tuni_emotion_full_metrics_outline_3sec.csv`:

| embedding | model | Val Balanced Accuracy |
|-----------|-------|-----------------------|
| whisper   | mlp   | 0.387                 |
| clap      | rf    | 0.370                 |


```python
from scripts.config.feature_selection_config import FeatureSelectionExperimentConfig, ModelEmbeddingPair
from scripts.config.partition_config import PartitionConfig
from scripts.config.experiment_config import ExperimentFilters
from scripts.run_feature_selection import run_feature_selection_experiment

from experiments.tuni_emotion.tuni_experiments_constants import EMOTIONS, TRAIN_SINGERS, VAL_SINGERS
from experiments.tuni_emotion.tuni_data_featureset_embedding_configs import (
    make_opensmile_embedding_config,
    make_whisper_embedding_config,
    make_clap_embedding_config,
)
```

```python
partition = PartitionConfig(
    name="tuni_emotion_singer_independent",
    train_data_ids=TRAIN_SINGERS,
    test_data_ids=VAL_SINGERS,
    data_column_label="singer",
)

opensmile_500ms = make_opensmile_embedding_config(
    parquet_file="tuni_emotion_dataset_opensmile-compare-2016_0.5s.parquet"
)   # ~6373 features
whisper_emb_500ms = make_whisper_embedding_config(
    parquet_file="tuni_emotion_dataset_whisper_whisper-base_0.5s.parquet"
)  # 512 features
clap_emb_500ms = make_clap_embedding_config(
    parquet_file="tuni_emotion_dataset_clap-2023_0.5s.parquet"
)         # 512 features
opensmile_3s = make_opensmile_embedding_config(
    parquet_file="tuni_emotion_dataset_opensmile-compare-2016_3.0s.parquet"
)  # ~6373 features
whisper_emb_3s = make_whisper_embedding_config(
    parquet_file="tuni_emotion_dataset_whisper_whisper-base_3.0s.parquet"
)  # 512 features
clap_emb_3s = make_clap_embedding_config(
    parquet_file="tuni_emotion_dataset_clap-2023_3.0s.parquet"
)  # 512 features

# ------------------------------------------------------------------
# EMBEDDING × MODEL PAIRS — pick based on the baseline results above
# ------------------------------------------------------------------
pairs_500ms = [
    ModelEmbeddingPair(dataset=opensmile_500ms, models=["svm"]),
    ModelEmbeddingPair(dataset=clap_emb_500ms, models=["svm"]),
]

pairs_3s = [
    ModelEmbeddingPair(dataset=whisper_emb_3s, models=["mlp"]),
    ModelEmbeddingPair(dataset=clap_emb_3s, models=["rf"]),
]

feature_selection_experiment_500ms = FeatureSelectionExperimentConfig(
    name="tuni_emotion_feature_selection_500ms",
    output_dir="./metrics",
    pairs=pairs_500ms,
    partition=partition,
    filters=ExperimentFilters(include_labels=EMOTIONS),
    frame_duration=0.5,  # TUNI embeddings use 0.5 s frames
    overlap=0.25,
    # Same hyper-parameters as the original trad_ml benchmark config
    model_params={
        "knn": {"n_neighbors": 5},
        "rf": {"n_estimators": 100, "random_state": 42},
        "mlp": {"hidden_layer_sizes": (12,), "max_iter": 500, "random_state": 42},
        "svm": {"kernel": "rbf", "C": 1.0},
    },
    feature_selection_methods=["rfe", "select_k_best"],
    n_features_list=[5, 10, 20, 40, 80, 160, 320, 640],
    seed=42,
)

feature_selection_experiment_3s = FeatureSelectionExperimentConfig(
    name="tuni_emotion_feature_selection_3sec",
    output_dir="./metrics",
    pairs=pairs_3s,
    partition=partition,
    filters=ExperimentFilters(include_labels=EMOTIONS),
    frame_duration=3.0,  # TUNI embeddings use 0.5 s frames
    overlap=0.25,
    # Same hyper-parameters as the original trad_ml benchmark config
    model_params={
        "knn": {"n_neighbors": 5},
        "rf": {"n_estimators": 100, "random_state": 42},
        "mlp": {"hidden_layer_sizes": (12,), "max_iter": 500, "random_state": 42},
        "svm": {"kernel": "rbf", "C": 1.0},
    },
    feature_selection_methods=["rfe", "select_k_best"],
    n_features_list=[5, 10, 20, 40, 80, 160, 320, 640],
    seed=42,
)
```

```python
results_500ms = run_feature_selection_experiment(feature_selection_experiment_500ms)
```

```python
results_3s = run_feature_selection_experiment(feature_selection_experiment_3s)
```
